```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
import numpy as np

output_dir = Path("../outputs")
output_dir.mkdir(parents=True, exist_ok=True)

results_path = output_dir / "day50_automated_quality_results.csv"
results_df = pd.read_csv(results_path)

print("Dataset shape:", results_df.shape)
display(results_df.head())
```

```python
snr_pivot = results_df.pivot(
    index="inactive_percentage",
    columns="method",
    values="SNR_dB"
)

snr_pivot = snr_pivot[["Incomplete", "Linear", "Cubic Spline"]]

ax = snr_pivot.plot(
    kind="bar",
    figsize=(9, 5),
    width=0.8
)

ax.set_xlabel("Inactive Sensors (%)")
ax.set_ylabel("SNR (dB)")
ax.set_title("SNR Comparison by Reconstruction Method")
ax.grid(axis="y", alpha=0.3)
ax.legend(title="Method")
plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig(
    output_dir / "day52_snr_bar_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()
```

```python
psnr_pivot = results_df.pivot(
    index="inactive_percentage",
    columns="method",
    values="PSNR_dB"
)

psnr_pivot = psnr_pivot[["Incomplete", "Linear", "Cubic Spline"]]

ax = psnr_pivot.plot(
    kind="bar",
    figsize=(9, 5),
    width=0.8
)

ax.set_xlabel("Inactive Sensors (%)")
ax.set_ylabel("PSNR (dB)")
ax.set_title("PSNR Comparison by Reconstruction Method")
ax.grid(axis="y", alpha=0.3)
ax.legend(title="Method")
plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig(
    output_dir / "day52_psnr_bar_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()
```

```python
ssim_pivot = results_df.pivot(
    index="inactive_percentage",
    columns="method",
    values="SSIM"
)

ssim_pivot = ssim_pivot[["Incomplete", "Linear", "Cubic Spline"]]

ax = ssim_pivot.plot(
    kind="bar",
    figsize=(9, 5),
    width=0.8
)

ax.set_xlabel("Inactive Sensors (%)")
ax.set_ylabel("SSIM")
ax.set_title("SSIM Comparison by Reconstruction Method")
ax.grid(axis="y", alpha=0.3)
ax.legend(title="Method")
plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig(
    output_dir / "day52_ssim_bar_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()
```

```python
snr_pivot.to_csv(output_dir / "day52_snr_comparison_table.csv")
psnr_pivot.to_csv(output_dir / "day52_psnr_comparison_table.csv")
ssim_pivot.to_csv(output_dir / "day52_ssim_comparison_table.csv")

print("All comparison tables saved successfully.")
```
