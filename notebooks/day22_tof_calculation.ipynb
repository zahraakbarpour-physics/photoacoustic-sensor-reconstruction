```python
import numpy as np

# Coordinates of a possible image point (meters)
x, y = 0.0, 0.0

# Coordinates of one sensor (meters)
sensor_x, sensor_y = 0.03, 0.04

# Speed of sound (meters per second)
c = 1500

# Calculate distance
distance = np.sqrt(
    (x - sensor_x)**2 +
    (y - sensor_y)**2
)

# Calculate time of flight
tof = distance / c

print(f"Distance: {distance * 1000:.2f} mm")
print(f"Time of flight: {tof * 1e6:.2f} microseconds")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 6))

plt.scatter(x, y, s=100, label="Image point")
plt.scatter(
    sensor_x, sensor_y,
    s=100, marker="^",
    label="Sensor"
)

plt.plot(
    [x, sensor_x],
    [y, sensor_y],
    linestyle="--",
    label=f"Distance = {distance * 1000:.0f} mm"
)

plt.xlabel("x (m)")
plt.ylabel("y (m)")
plt.title("Sensor-to-Point Geometry")
plt.axis("equal")
plt.grid(True)
plt.legend()

from pathlib import Path

from pathlib import Path

output_dir = Path("../outputs")
output_dir.mkdir(parents=True, exist_ok=True)

plt.savefig(
    output_dir / "day22_sensor_point_geometry.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.show()
plt.show()
```
