```python
import numpy as np
import matplotlib.pyplot as plt

# Signal values extracted at the predicted arrival times
signals = np.array([4, 5, 6, 5])

# Delay-and-Sum for one image point
reconstructed_value = np.sum(signals)

print("Sensor signals:", signals)
print("Reconstructed value:", reconstructed_value)
```

```python
from pathlib import Path

plt.figure(figsize=(7, 4))

sensor_numbers = np.arange(1, len(signals) + 1)

plt.bar(sensor_numbers, signals)

plt.xlabel("Sensor Number")
plt.ylabel("Signal Amplitude")
plt.title("Signal Contributions to One DAS Pixel")
plt.xticks(sensor_numbers)
plt.grid(axis="y", alpha=0.3)

output_dir = Path("../outputs")
output_dir.mkdir(parents=True, exist_ok=True)

plt.savefig(
    output_dir / "day23_das_signal_contributions.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
