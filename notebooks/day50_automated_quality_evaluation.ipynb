```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import json

from pathlib import Path
from scipy.interpolate import interp1d, CubicSpline
from skimage.metrics import structural_similarity

print("All libraries imported successfully!")
```

```python
def calculate_snr(reference, estimate):
    reference = np.asarray(reference, dtype=float)
    estimate = np.asarray(estimate, dtype=float)

    if reference.shape != estimate.shape:
        raise ValueError("Image shapes must match.")

    error_power = np.sum((reference - estimate) ** 2)
    signal_power = np.sum(reference ** 2)

    if error_power == 0:
        return float("inf")
    if signal_power == 0:
        raise ValueError("Reference signal power is zero.")

    return float(10 * np.log10(signal_power / error_power))


def calculate_psnr(reference, estimate):
    reference = np.asarray(reference, dtype=float)
    estimate = np.asarray(estimate, dtype=float)

    if reference.shape != estimate.shape:
        raise ValueError("Image shapes must match.")

    data_range = np.max(reference) - np.min(reference)

    if data_range <= 0:
        raise ValueError("Reference image has no intensity range.")

    mse = np.mean((reference - estimate) ** 2)

    if mse == 0:
        return float("inf")

    return float(10 * np.log10(data_range**2 / mse))


def calculate_ssim(reference, estimate):
    reference = np.asarray(reference, dtype=np.float64)
    estimate = np.asarray(estimate, dtype=np.float64)

    if reference.shape != estimate.shape:
        raise ValueError("Image shapes must match.")

    data_range = float(np.max(reference) - np.min(reference))

    if data_range <= 0:
        raise ValueError("Reference image has no intensity range.")

    min_dim = min(reference.shape)
    win_size = min(7, min_dim if min_dim % 2 == 1 else min_dim - 1)

    if win_size < 3:
        raise ValueError("Image is too small for SSIM.")

    return float(
        structural_similarity(
            reference,
            estimate,
            data_range=data_range,
            win_size=win_size
        )
    )


def evaluate_image(reference, estimate):
    return {
        "SNR_dB": calculate_snr(reference, estimate),
        "PSNR_dB": calculate_psnr(reference, estimate),
        "SSIM": calculate_ssim(reference, estimate)
    }


print("All three evaluation metrics are ready.")
```

```python
num_sensors = 32
num_samples = 600

fs = 10e6
dt = 1 / fs
c = 1500.0

source_x = 5e-3
source_y = 20e-3
pulse_width = 1.5e-6

sensor_x = np.linspace(-15e-3, 15e-3, num_sensors)
time = np.arange(num_samples) * dt

signals = np.zeros((num_sensors, num_samples))

for i in range(num_sensors):
    distance = np.sqrt(
        (source_x - sensor_x[i])**2 + source_y**2
    )
    arrival_time = distance / c

    signals[i] = np.exp(
        -0.5 * ((time - arrival_time) / pulse_width)**2
    )

x_grid = np.linspace(-10e-3, 10e-3, 100)
y_grid = np.linspace(2e-3, 30e-3, 100)

X, Y = np.meshgrid(x_grid, y_grid)


def reconstruct_das(signal_matrix):
    image = np.zeros_like(X, dtype=float)

    for i in range(num_sensors):
        distance_map = np.sqrt(
            (X - sensor_x[i])**2 + Y**2
        )

        indices = np.rint(
            (distance_map / c) / dt
        ).astype(int)

        valid = (indices >= 0) & (indices < num_samples)

        sampled = np.zeros_like(X, dtype=float)
        sampled[valid] = signal_matrix[i, indices[valid]]

        image += sampled

    return image


reference_image = reconstruct_das(signals)

print("Signal matrix shape:", signals.shape)
print("Reference image shape:", reference_image.shape)
```

```python
inactive_counts = [0, 4, 8, 12, 16]
dropout_order = np.random.default_rng(42).permutation(num_sensors)

results = []

for n_inactive in inactive_counts:

    inactive_indices = np.sort(dropout_order[:n_inactive])
    active_indices = np.setdiff1d(
        np.arange(num_sensors),
        inactive_indices
    )

    # Signals with inactive sensors set to zero
    incomplete_signals = signals.copy()
    incomplete_signals[inactive_indices, :] = 0

    # Reconstruct missing sensor signals
    linear_signals = incomplete_signals.copy()
    spline_signals = incomplete_signals.copy()

    if n_inactive > 0:
        for sample_idx in range(num_samples):

            active_values = signals[active_indices, sample_idx]

            linear_function = interp1d(
                sensor_x[active_indices],
                active_values,
                kind="linear",
                bounds_error=False,
                fill_value="extrapolate"
            )

            spline_function = CubicSpline(
                sensor_x[active_indices],
                active_values,
                extrapolate=True
            )

            linear_signals[inactive_indices, sample_idx] = (
                linear_function(sensor_x[inactive_indices])
            )

            spline_signals[inactive_indices, sample_idx] = (
                spline_function(sensor_x[inactive_indices])
            )

    # DAS reconstruction
    incomplete_image = reconstruct_das(incomplete_signals)
    linear_image = reconstruct_das(linear_signals)
    spline_image = reconstruct_das(spline_signals)

    estimates = {
        "Incomplete": incomplete_image,
        "Linear": linear_image,
        "Cubic Spline": spline_image
    }

    # Evaluate each reconstructed image against the reference
    for method, estimate in estimates.items():

        metrics = evaluate_image(reference_image, estimate)

        results.append({
            "inactive_sensors": int(n_inactive),
            "inactive_percentage": round(
                100 * n_inactive / num_sensors, 2
            ),
            "method": method,
            **metrics
        })

    print(f"Completed: {n_inactive} inactive sensors")


results_df = pd.DataFrame(results)

print("\nAutomated evaluation results:")
display(results_df.round(4))
```

```python
output_dir = Path("../outputs")
output_dir.mkdir(parents=True, exist_ok=True)

csv_path = output_dir / "day50_automated_quality_results.csv"
excel_path = output_dir / "day50_automated_quality_results.xlsx"
json_path = output_dir / "day50_automated_quality_results.json"

results_df.to_csv(csv_path, index=False)
results_df.to_excel(excel_path, index=False)
results_df.to_json(
    json_path,
    orient="records",
    indent=4
)

print("Saved files:")
print(csv_path)
print(excel_path)
print(json_path)
```

```python
from pathlib import Path

output_dir = Path("../outputs")
output_dir.mkdir(parents=True, exist_ok=True)

csv_path = output_dir / "day50_automated_quality_results.csv"
excel_path = output_dir / "day50_automated_quality_results.xlsx"
json_path = output_dir / "day50_automated_quality_results.json"

results_df.to_csv(csv_path, index=False)
results_df.to_excel(excel_path, index=False)
results_df.to_json(json_path, orient="records", indent=4)

print("All result files saved successfully!")
```

```python
metrics_to_plot = [
    ("SNR_dB", "SNR (dB)", "day50_snr_automated.png"),
    ("PSNR_dB", "PSNR (dB)", "day50_psnr_automated.png"),
    ("SSIM", "SSIM", "day50_ssim_automated.png")
]

for metric, ylabel, filename in metrics_to_plot:
    plt.figure(figsize=(8, 5))

    for method in ["Incomplete", "Linear", "Cubic Spline"]:
        subset = results_df[results_df["method"] == method]

        plt.plot(
            subset["inactive_percentage"],
            subset[metric],
            marker="o",
            label=method
        )

    plt.xlabel("Inactive sensors (%)")
    plt.ylabel(ylabel)
    plt.title(f"{ylabel} vs. Inactive Sensor Percentage")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()

    plt.savefig(
        output_dir / filename,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()

print("All plots saved successfully!")
```
