```python
import sys
print(sys.executable)

import numpy as np
import matplotlib.pyplot as plt

print("Matplotlib is ready!")
```

```python
t = np.linspace(0, 1, 500)
signal = np.sin(2 * np.pi * 5 * t)

plt.figure(figsize=(10, 4))
plt.plot(t, signal, label="Sensor signal")

plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title("Simulated Photoacoustic Sensor Signal")

plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig("../outputs/day05_original_signal.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
noise = 0.2 * np.random.randn(len(t))
noisy_signal = signal + noise

plt.figure(figsize=(10, 4))

plt.plot(t, signal, label="Original signal")
plt.plot(t, noisy_signal, label="Noisy signal", alpha=0.7)

plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title("Original vs Noisy Signal")

plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig("day05_noisy_signal.png", dpi=300, bbox_inches="tight")
plt.show()
```
